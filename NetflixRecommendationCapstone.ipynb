## Project Summary

This project focuses on building a movie recommendation system using the Netflix dataset. The primary objective is to identify popular genres, build a collaborative filtering model, and provide personalized movie recommendations.

The key steps involved are:

1.  **Data Loading and Initial Exploration**:
    *   The `combined_data_1.txt` file, containing customer ratings, was loaded into a Pandas DataFrame (`df_User_Id`).
    *   Initial checks were performed for null values and to understand the distribution of ratings.
    *   The number of movies and customers were calculated, and the distribution of ratings (1-5 stars) was analyzed.

2.  **Data Preprocessing and Feature Engineering**:
    *   A `movie_Id` column was extracted from the `Cust_Id` column in `df_User_Id` to link ratings to movies.
    *   Null ratings were removed from `df_User_Id`.
    *   Movie title information was loaded from `movies (1) (1).csv` into `df_titles`, and the header row was dropped. The `movie_Id` column in `df_titles` was converted to a numeric type.
    *   The `df_User_Id` and `df_titles` DataFrames were merged on `movie_Id` to create a comprehensive `df` containing customer ratings, movie titles, and genres.

3.  **Filtering for Popular and Well-Rated Content**:
    *   The DataFrame `df` was filtered to `df_pop` to include only ratings greater than or equal to 4, aiming to focus on highly-rated movies.
    *   Benchmarks were established to filter out users and movies with insufficient activity/ratings:
        *   Users with fewer ratings than the 75th percentile of user ratings (`user_benchmark = 38`) were dropped.
        *   Movies with fewer ratings than the 70th percentile of movie ratings (`movie_benchmark = 840`) were dropped.
    *   This filtering helps to reduce noise and focus on more reliable data for recommendation.

4.  **Model Building (Collaborative Filtering)**:
    *   The `scikit-surprise` library was used to build a collaborative filtering model.
    *   A `Reader` object was initialized, and a `Dataset` was created from the `Cust_Id`, `movie_Id`, and `Rating` columns of the filtered `df_pop` DataFrame.
    *   A Singular Value Decomposition (SVD) model was initialized and evaluated using 4-fold cross-validation with RMSE and MAE metrics.

5.  **Generating Recommendations**:
    *   Recommendations were generated for a specific user (customer ID '15737').
    *   The SVD model predicted an 'Estimated_Score' for movies this user had rated. The top 5 movies with the highest estimated scores for this user were identified.

6.  **Genre Analysis**:
    *   The analysis identified 'Drama' as the most popular and liked genre based on `movie_Id` counts in highly-rated movies.
    *   The notebook also displayed movies and genres that received best (ratings >= 4) and worst (ratings <= 2) ratings, highlighting commonalities and differences in user preferences across these categories.


# Conclusion:
From the above analysis, it is observed that certain movies and genres appear in both the best-rated and worst-rated categories, which indicates that user perception varies depending on the number and diversity of users rating the same content.

The dataset shows that more than 1 crore records have ratings greater than or equal to 4, suggesting that a large volume of users have positively rated many movies. In contrast, approximately 35.16 lakh records have ratings below 3, indicating that comparatively fewer users have given low ratings.

This imbalance highlights that highly rated movies are supported by stronger user participation, whereas poorly rated movies tend to have limited engagement. Additionally, some movies appear exclusively in either the high-rating or low-rating groups, implying that not all content overlaps across these filtered segments and that user preference is not uniformly distributed across all movies and genres.
